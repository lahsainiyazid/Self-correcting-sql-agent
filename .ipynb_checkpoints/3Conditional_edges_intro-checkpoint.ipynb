```python
from typing import TypedDict,Literal
from langgraph.graph import StateGraph,END
class RAGState(TypedDict):
    #We define here our sharedstate:
    query:str
    documents:list[str]
    is_relevant:bool    
#We define our nodes (Pure fonctions)
def retrieve_docs(state:RAGState)->dict:
    return {"documents":["LangGraph allows dynamic routing via conditional edges."]}
def grade_docs(state:RAGState)->dict:
    #Simulating LLM grading the documents:
    return{"is_relevant":True}
def route_after_grading(state:RAGState)->Literal["generate","rewrite"]:
    if state["is_relevant"]:
        return "generate"
    return "rewrite"
def generate_answer(state:RAGState)->dict:
    return {"Query":"Answer generated successfully!"}
def rewrite_query(state:RAGState)->dict:
    return {"query":"Revised query based on poor context."}
```

Conditional edges allow our graph to dynamically choose the next node at run time:

```python
#Now we build the graph:
workflow=StateGraph(RAGState)
#We add our nodes:
workflow.add_node("retrieve",retrieve_docs)
workflow.add_node("grade",grade_docs)
workflow.add_node("generate",generate_answer)
workflow.add_node("rewrite",rewrite_query)
#Set up explicit linear edges:
workflow.set_entry_point("retrieve")
workflow.add_edge("retrieve","grade")
workflow.add_conditional_edges("grade", #Source node
                              route_after_grading, #Inspects condition:
                               
                               #If the node returns generate we execute the generate node
                               #If it returns rewrite we execute the rewrite node
                              {"generate":"generate",
                              "rewrite":"rewrite"})
app=workflow.compile()
```

Stategraph->Executable flowchart it contains:

State->DS (Typeddict) serves as ground truth\


Nodes->Py functions (takes states->performs operations->updates states)

Edges->Roads connecting to nodes ,defining execution path

Ps:The execution flowchart is not linear from left to right like for langchain!

```python
'''
We use typedict so that our object of type typeddict and hence accepted by langgraph!
'''
from typing import TypedDict
class GraphState(TypedDict):#Our class inherits of Typeddict
    user_message:str
    processed_count:int
def process_message_node(state:GraphState)->GraphState:
    print(f"Processing{state['user_message']}")
    return {"user_message":state["user_message"].upper()+"!",
            "processed_count":state["processed_count"]+1}
```

```python
from langgraph.graph import StateGraph,END
workflow=StateGraph(GraphState)
workflow.add_node("processor_node",process_message_node)
workflow.set_entry_point("processor_node")
workflow.add_edge("processor_node",END)
app=workflow.compile()
```

### **Summary: Tool Binding in LangGraph**

* **What it is:** Tool binding is the process of attaching local Python function schemas (`@tool`) to an LLM using `.bind_tools()`. This gives the model a **"menu"** of available functions it can request.
* **How it works (The Division of Roles):**
1. **The LLM (The Decision Maker / Manager):**
* Never executes Python code directly—it has zero access to the local machine or server.
* Acts as an **intent router** or **judge**: reads the prompt, inspects tool metadata (names, descriptions, docstrings), and determines *if* a tool is needed, *which* tool fits, and *what arguments* to pass.
* Returns a structured request (`tool_calls` JSON) specifying the function name and parameters.


2. **Local Code / LangGraph (The Executor / Specialist):**
* Listens for the LLM's `tool_calls` request via nodes like `ToolNode`.
* Actually runs the Python function locally (executing math, SQL queries, API calls, file I/O).
* Passes the result back to the LLM in a `ToolMessage` so the model can formulate a final answer.




* **When No Tool Fits:**
* In standard `tool_choice="auto"` mode, if no tool matches the task, the LLM bypasses tool calling completely and responds directly with standard text (or politely explains its limitations).

ToolNode acts as the Specialist / Executor. It is a pre-built LangGraph component that listens for the LLM's request to use a tool, actually runs the local Python function, and hands the result back to the LLM. 

```python
from dotenv import load_dotenv
from typing import TypedDict,Annotated #We add annotated to be able to use add to update the state
from operator import add 
from langchain_groq import ChatGroq
from langchain_core.tools import tool
from langgraph.graph import StateGraph,END
from langgraph.prebuilt import ToolNode
load_dotenv()
#1 Define the state schema:
class AgentState(TypedDict):
    messages:Annotated[list,add] #list->type hint ,add ->operator used to update the state!
#2 Define the tool:
@tool
def search_tool(query:str)->str:
    """
    Searching info on the web.
    """
    return f"Resarch results for {query}"
#3 Initialize our llm model:
llm=ChatGroq(model="llama-3,3-70b-versatile",temperature=0)
llm_with_tool=llm.bind_tools([search_tool])
#4 Define node functions:
def call_model_with_tools(state:AgentState):
    messages=state["messages"]
    response=llm_with_tools.invoke(messages)
    return {"messages":[response]}
def should_continue(state:AgentState)->str:
    last_msg=state["messages"][-1]
    if hasattr(last_msg,"tool_calls") and last_msg.tool_calls:
        return "tools"
    return "end"
#5 Build and compile the StateGraph:
workflow=StateGraph(AgentState)
workflow.add_node("agent",call_model_with_tools)
workflow.add_node("tools",ToolNode(tools=[search_tool]))
workflow.set_entry_point("agent")
workflow.add_conditional_edges("agent",
                              should_continue,
                              {"tools":"tools",
                              "end":END})
workflow.add_edge("tools","agent")
app=workflow.compile()
```
